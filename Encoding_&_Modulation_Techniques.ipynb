```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
bitstream = [0, 1, 1, 0, 1, 0, 0, 1]   # Example digital input bits
fs = 1000                              # Sampling frequency
Tb = 1                                 # Bit duration (1s for simplicity)
N = len(bitstream)
t = np.arange(0, N*Tb, 1/fs)

# FSK parameters
f0, f1 = 5, 10   # Frequencies for bit 0 and 1

# Generate digital signal
bit_signal = np.repeat(bitstream, fs*Tb)

# Carrier signal (use f0 just for reference)
carrier = np.cos(2 * np.pi * f0 * t)

# Generate FSK signal
fsk_signal = np.array([])
for bit in bitstream:
    tt = np.arange(0, Tb, 1/fs)
    if bit == 0:
        wave = np.cos(2*np.pi*f0*tt)
    else:
        wave = np.cos(2*np.pi*f1*tt)
    fsk_signal = np.concatenate((fsk_signal, wave))

# Plotting
plt.figure(figsize=(12,8))

# Digital signal
plt.subplot(3,1,1)
plt.plot(t, bit_signal, 'b')
plt.title('Digital Signal (Bitstream)')
plt.ylabel('Amplitude')
plt.grid(True)

# Carrier signal
plt.subplot(3,1,2)
plt.plot(t, carrier, 'g')
plt.title('Carrier Wave (Reference)')
plt.ylabel('Amplitude')
plt.grid(True)

# FSK modulated signal
plt.subplot(3,1,3)
plt.plot(t, fsk_signal, 'r')
plt.title('Frequency Shift Keying (FSK)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
bitstream = [1, 0, 1, 1, 1, 0, 0, 1]    # Example input bits
fc = 5                                 # Carrier frequency in Hz
fs = 1000                              # Sampling frequency
Tb = 1                                 # Bit duration (1s for simplicity)
N = len(bitstream)
t = np.arange(0, N*Tb, 1/fs)

# Generate digital signal
bit_signal = np.repeat(bitstream, fs*Tb)

# Carrier wave
carrier = np.cos(2 * np.pi * fc * t)

# Generate ASK signal (bit=1 → carrier, bit=0 → 0V)
ask_signal = np.array([])
for bit in bitstream:
    tt = np.arange(0, Tb, 1/fs)
    if bit == 1:
        wave = np.cos(2*np.pi*fc*tt)   # Carrier present
    else:
        wave = np.zeros_like(tt)       # 0V for bit=0
    ask_signal = np.concatenate((ask_signal, wave))

# Plotting
plt.figure(figsize=(12,8))

# Digital signal
plt.subplot(3,1,1)
plt.plot(t, bit_signal, 'b')
plt.title('Digital Signal (Bitstream)')
plt.ylabel('Amplitude')
plt.grid(True)

# Carrier wave
plt.subplot(3,1,2)
plt.plot(t, carrier, 'g')
plt.title('Carrier Wave')
plt.ylabel('Amplitude')
plt.grid(True)

# ASK modulated signal
plt.subplot(3,1,3)
plt.plot(t, ask_signal, 'r')
plt.title('Amplitude Shift Keying (ASK) [0-bit → 0V]')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
bitstream = [1, 0, 1, 1, 1, 0, 0, 1]   # Example bitstream
fc = 4                                 # Carrier frequency (Hz)
fs = 1000                              # Sampling frequency
Tb = 1.0                               # Bit duration (sec)

N = len(bitstream)
samples_per_bit = int(fs * Tb)
t = np.arange(0, N*Tb, 1/fs)

# Digital signal (for plotting)
bit_signal = np.repeat(bitstream, samples_per_bit)

# Base carrier across the entire duration (continuous)
carrier = np.cos(2 * np.pi * fc * t)

# Build BPSK: flip by 180° only on bit change
flips = np.ones(N)
for k in range(1, N):
    if bitstream[k] != bitstream[k-1]:
        flips[k] = -flips[k-1]   # toggle sign on bit change (mirror)
    else:
        flips[k] = flips[k-1]    # keep same sign otherwise

flip_signal = np.repeat(flips, samples_per_bit)

# BPSK modulated wave
bpsk_signal = carrier * flip_signal

# ===== Plotting =====
plt.figure(figsize=(12,9))

# Digital signal
plt.subplot(3,1,1)
plt.plot(t, bit_signal, 'b')
plt.title('Digital Signal (Bitstream)')
plt.ylim(-0.5, 1.5)
plt.ylabel('Bit Level')
plt.grid(True)
for i, b in enumerate(bitstream):
    plt.axvline(i*Tb, linestyle='--', alpha=0.3)
    plt.text(i*Tb + Tb/2, 1.2, str(b), ha='center')

# Carrier wave
plt.subplot(3,1,2)
plt.plot(t, carrier, 'g')
plt.title(f'Carrier Wave (fc = {fc} Hz)')
plt.ylabel('Amplitude')
plt.grid(True)

# BPSK wave
plt.subplot(3,1,3)
plt.plot(t, bpsk_signal, 'r')
plt.title('PSK Signal')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
for i in range(N+1):
    plt.axvline(i*Tb, linestyle='--', alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
Am = 1          # Message amplitude
fm = 2          # Message frequency (Hz) - low frequency
Ac = 1          # Carrier amplitude
fc = 20         # Carrier frequency (Hz) - higher than message
Fs = 1000       # Sampling frequency
mod_index = 0.8 # Modulation index (0 < m <= 1)

# Time axis
t = np.arange(0, 1, 1/Fs)  # 1 second duration

# Message signal
message = Am * np.sin(2 * np.pi * fm * t)

# Carrier signal
carrier = Ac * np.cos(2 * np.pi * fc * t)

# AM signal: s(t) = (1 + m * sin(...)) * cos(...)
am_signal = (1 + mod_index * np.sin(2 * np.pi * fm * t)) * np.cos(2 * np.pi * fc * t)

# Plot
plt.figure(figsize=(10,8))

# (i) Message signal
plt.subplot(3,1,1)
plt.plot(t, message, color='green')
plt.title('(i) Message Signal (Low Frequency)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (ii) Carrier signal
plt.subplot(3,1,2)
plt.plot(t, carrier, color='orange')
plt.title('(ii) Carrier Signal (High Frequency)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (iii) AM signal
plt.subplot(3,1,3)
plt.plot(t, am_signal, color='blue')
plt.title('(iii) Amplitude Modulated (AM) Signal')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
Am = 1          # Message amplitude
fm = 2          # Message frequency (Hz) - low frequency
Ac = 1          # Carrier amplitude
fc = 20         # Carrier frequency (Hz) - higher than message
Fs = 1000       # Sampling frequency
beta = 5        # Frequency deviation index

# Time axis
t = np.arange(0, 1, 1/Fs)  # 1 second duration

# Message signal
message = Am * np.sin(2 * np.pi * fm * t)

# Carrier signal
carrier = Ac * np.cos(2 * np.pi * fc * t)

# FM signal
fm_signal = Ac * np.cos(2 * np.pi * fc * t + beta * np.sin(2 * np.pi * fm * t))

# Plot
plt.figure(figsize=(10,8))

# (i) Message signal
plt.subplot(3,1,1)
plt.plot(t, message, color='green')
plt.title('(i) Message Signal (Low Frequency)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (ii) Carrier signal
plt.subplot(3,1,2)
plt.plot(t, carrier, color='orange')
plt.title('(ii) Carrier Signal (High Frequency)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (iii) FM signal
plt.subplot(3,1,3)
plt.plot(t, fm_signal, color='blue')
plt.title('(iii) Frequency Modulated (FM) Signal')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
Am = 1          # Message amplitude
fm = 2          # Message frequency (Hz) - analog signal (low freq)
Ac = 1          # Carrier amplitude
fc = 30         # Carrier frequency (Hz) - much higher than message
Fs = 2000       # Sampling frequency
kp = np.pi/2    # Phase deviation constant (rad per unit amplitude)

# Time axis
t = np.arange(0, 1, 1/Fs)  # simulate 1 sec

# Analog message signal (sine wave)
message = Am * np.sin(2 * np.pi * fm * t)

# Carrier signal (cosine wave)
carrier = Ac * np.cos(2 * np.pi * fc * t)

# Phase Modulated signal (Analog-to-Analog conversion)
pm_signal = Ac * np.cos(2 * np.pi * fc * t + kp * message)

# Plotting
plt.figure(figsize=(12,8))

# (i) Message (Analog Input Signal)
plt.subplot(3,1,1)
plt.plot(t, message, 'g')
plt.title('(i) Analog Message Signal (m(t))')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (ii) Carrier Signal
plt.subplot(3,1,2)
plt.plot(t, carrier, 'orange')
plt.title('(ii) Carrier Signal (c(t))')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

# (iii) Phase Modulated Signal
plt.subplot(3,1,3)
plt.plot(t, pm_signal, 'b')
plt.title('(iii) Phase Modulated Signal (PM)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)

plt.tight_layout()
plt.show()
```
